```python
import pandas as pd
from io import StringIO

with open("athlete_events.csv", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

contenido = "\n".join(
    linea[1:-1].replace('""', '"')
    for linea in contenido.splitlines()
)

datos = pd.read_csv(StringIO(contenido))

datos.head()
```

```python
faltantes = datos.isnull().sum()

porcentaje = (faltantes / len(datos)) * 100

resumen = pd.DataFrame({
    "Datos vacíos": faltantes,
    "Porcentaje": porcentaje
})

resumen
```

```python
datos = datos.drop_duplicates()

print("Tamaño de la base limpia:", datos.shape)
```

```python
datos.to_csv("datos.csv", index=False)
```
